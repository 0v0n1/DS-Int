```python
#while loop to compute factorial of a user inputted number
n=int(input("Enter the number: "))
i=1
fac=1
while i<=n: #while i!=0:
    fac*=i
    i+=1 #i-=1
print(fac)
```

```python
#list of vowels from a sentence
v=["a","e","i","o","u","A","E","I","O","U"]
s=input("enter a sentence: ")
l=[]
for i in s:
    if i in v:
        l.append(i)
print(l)
```

```python
#using list comprehension
s=input("enter a sentence: ")
v="aeiouAEIOU"
filter=[char for char in s if char in v]
print(filter)
```

```python
def info(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")
info(name="avani", age=21, city="lko")
```
